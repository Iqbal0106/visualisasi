Import Library

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import plotly.graph_objects as go

# Membuat DataFrame dari data contoh
data = {
    "Partisipan": np.arange(1, 11),
    "Peng Makanan": [50, 40, 60, 55, 45, 65, 70, 75, 80, 990],
    "Peng Transportasi": [20, 25, 30, 35, 40, 50, 55, 60, 65, 70],
    "Berat Badan": [60, 55, 65, 70, 62, 75, 80, 85, 90, 95]
}
df = pd.DataFrame(data)
```

Penentuan Variabel dan Estimasi Model OLS

```python
X = df[["Peng Makanan", "Peng Transportasi"]]
y = df["Berat Badan"]

X_model = sm.add_constant(X)

model = sm.OLS(y, X_model)
results = model.fit()

print("=" * 70)
print("HASIL REGRESI LINEAR BERGANDA")
print("=" * 70)
print(results.summary())
```

Membentuk persamaan regresi dan koefisien

```python
intercept = results.params["const"]
coef_makanan = results.params["Peng Makanan"]
coef_transportasi = results.params["Peng Transportasi"]
r_squared = results.rsquared
adjusted_r_squared = results.rsquared_adj

persamaan = (
    f"Y = {intercept:.3f}"
    f" + ({coef_makanan:.3f}) X1"
    f" + ({coef_transportasi:.3f}) X2"
)

print("=" * 70)
print("PERSAMAAN REGRESI")
print("=" * 70)
print(persamaan)
print(f"R² = {r_squared:.4f}")
print(f"Adjusted R² = {adjusted_r_squared:.4f}")
```

Perhitungan Prediksi Nilai Y

```python
df["Prediksi"] = results.predict(X_model)

print("=" * 70)
print("DATA AKTUAL & PREDIKSI")
print("=" * 70)
print(df[["Partisipan", "Peng Makanan", "Peng Transportasi", "Berat Badan", "Prediksi"]].round(2))
```

```python
# Membuat Grid untuk Bidang Regresi
x1 = np.linspace(df["Peng Makanan"].min(), df["Peng Makanan"].max(), 40)
x2 = np.linspace(df["Peng Transportasi"].min(), df["Peng Transportasi"].max(), 40)
X1, X2 = np.meshgrid(x1, x2)
Y_plane = (intercept + coef_makanan * X1 + coef_transportasi * X2)

# Membuat Figure Plotly 3D
fig = go.Figure()

# Menampilkan Data Aktual (Scatter Plot)
fig.add_trace(go.Scatter3d(
    x=df["Peng Makanan"], y=df["Peng Transportasi"], z=df["Berat Badan"],
    mode="markers", marker=dict(size=7, color="red", opacity=0.9),
    name="Data Aktual"
))

# Menampilkan Bidang Regresi (Surface Plot)
fig.add_trace(go.Surface(
    x=X1, y=X2, z=Y_plane, opacity=0.60, colorscale="Viridis",
    name="Regression Plane"
))

# Pengaturan Layout Grafik
fig.update_layout(
    title={"text": f"REGRESI LINEAR BERGANDA 3D<br><sup>{persamaan} | R² = {r_squared:.3f}</sup>", "x": 0.5},
    scene={
        "xaxis": {"title": "Pengeluaran Makanan (X1)"},
        "yaxis": {"title": "Pengeluaran Transportasi (X2)"},
        "zaxis": {"title": "Berat Badan (Y)"}
    }
)

fig.show()
```
